```python
type(df['seizures_aryan'])
```

```python
checked=df[(df['seizures_aryan']=='1') | (df['seizures_aryan']=='0')]
checked=checked[checked['llm_seizure'].notnull()]
checked['llm_seizure']=checked['llm_seizure'].astype(int)
checked['seizures_aryan']=checked['seizures_aryan'].astype(int)
```

```python
checked['seizures_aryan'].value_counts()
```

```python
checked['llm_seizure'].value_counts()
```

```python

checked['llm_sz_correct']=checked['llm_seizure'].values*checked['seizures_aryan'].values
checked['llm_sz_correct'].sum()
```

```python
checked['false_positives']= (checked['llm_seizure']==1) & (checked['seizures_aryan']==0)
checked['false_negatives']= (checked['llm_seizure']==0) & (checked['seizures_aryan']==1)
print(checked['false_positives'].sum(),checked['false_negatives'].sum())
```
