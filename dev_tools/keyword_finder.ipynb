```python
import pandas as pd
import os
```

```python
df=pd.read_csv("/Users/rm026/Documents/schmahmann/ground_truth_master_list_ccas_partial.csv")
root_folder="/Users/rm026/Documents/schmahmann/schmahmann_output_w_unknown/_separated/"
keywords=['phonem','fluen','categor','vegetabl','verbal con','verbal','registr','digit ','digits','digit,','digit.','drawn','draw','cube','similarit','go-no','no-go','affect','labile','apath','irritab']
finds=[]
for file in os.listdir(root_folder):
    if not file.endswith('.txt'):
        continue
    with open(os.path.join(root_folder,file),'r') as f:
        text=f.read().lower()
        mrn=int(file.split('.')[0])
        in_file=''
        for kw in keywords:
            kw_count=text.lower().count(kw)
            if kw_count!=0:
                in_file+=f"{kw}:{kw_count}; "
        finds.append([mrn,in_file])
finds_df=pd.DataFrame(finds, columns=['MRN','keywords_counts_lower'])
finds_df
```

```python
merged=pd.merge(df,finds_df,on='MRN')
merged.to_csv("/Users/rm026/Documents/schmahmann/ground_truth_master_list_ccas_partial.csv")
```
