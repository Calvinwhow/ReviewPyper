```python
import pandas as pd
import os
```

```python
scores=pd.read_csv("/Users/rm026/Partners HealthCare Dropbox/Ross Macfadyen/Schmahmann_SCA_Atrophy/schmahmann_connectivity_filelist.csv")
scores=scores[['mrn']+[col for col in scores.columns if col.startswith('Sec')]]

df=pd.read_csv("/Users/rm026/Documents/schmahmann/ground_truth_cnrs_schmahmann_wip.csv")
new_names={df.columns[2*i+1]:df.columns[2*i+1]+': '+colname[5:] for i, colname in enumerate(scores.columns[1:])}
scores.columns=['MRN'] + list(new_names.values())
new_names.update({f'explanation: {old}':f'explanation: {new}' for old, new in new_names.items()})
df.rename(columns=new_names, inplace=True)
df
```

```python
merge=pd.merge(df, scores, on='MRN',suffixes=['_me','_sch'], how='left')
```

```python

import numpy as np

check_col=[]
for i, row in merge.iterrows():
    check=[]

    for col in scores.columns[1:]:

        if int(row[col+'_me'])==2:# if I already checked, skip
            continue 

        if np.isnan(row[col+'_sch']) or row[col+'_sch']!=0: # if I say no/unknown and schmahmann finds something
            check.append(col)


    check_col.append(check)

merge['needs_checking']=check_col
merge
```

```python

import json
root_folder="/Users/rm026/Documents/schmahmann/schmahmann_cnrs_and_similarities_redo/_separated/"

# keywords=['phonem','fluen','categor','vegetabl','verbal con','verbal','registr','digit ','digits','digit,','digit.','drawn','draw','cube','similarit','go-no','no-go','affect','labile','apath','irritab']
# keywords={'attention a':['attention', 'focus','distract', 'restless', 'compel', 'compuls', 'habit', 'repetitive', 'inflexib', 'worry'], 
#           'attention b':['anxi', 'inappropri', ], 
#           'emotion a':['interrupt', 'emotion', 'worry','laugh',  'fear', 'panic'], 
#           'emotion b':['pleasure', 'enjoy', 'pessim', 'hopeless','dissatis', 'depress', 'sad',], 
#           'autism a':[ 'sensory', 'repetit', 'habit'], 
#           'autism b':['sensory','sensitiv', 'overwhelm', 'overstimul'],
#           'psychosis a':['illogical', 'inappropri','rambling', 'schizo', 'irrational', 'paranoi', 'hallucinat'],
#           'psychosis b':['interaction', 'aloof', 'affect', 'reactivity', 'indifferen'],
#           'social a':['anger', 'angry', 'aggress', 'irrita', 'argument','defiant','intoler', 'protective'], 
#           'social b':['appropri', 'trustful', 'trusting', 'trick', 'manipulat']
#           }
# keywords=keywords.split()
keywords=json.load(open('cnrs_keywords.json'))
keywords={new_names[section]:words.split(', ') for words, section in keywords.items()}
keywords
```

```python
finds=[]
to_check=[]
for file in os.listdir(root_folder):

    if not file.endswith('.txt'):
        continue
    
    mrn=int(file.split('.')[0])
    row=merge[merge['MRN']==mrn]
    cols_to_check=row['needs_checking']

    with open(os.path.join(root_folder,file),'r') as f:

        text=f.read().lower()

        in_file=''

        for q, wordlist in keywords.items():
            
            if q not in cols_to_check.iloc[0]:

                continue

            text_out=q+': '
            
            found=False
            for word in wordlist:

                kw_count=text.lower().count(word)

                if kw_count!=0:
                    found=True
                    text_out+=f"{word}:{kw_count},"
            if found:
                in_file+=text_out+'\n'
            
        finds.append([mrn,in_file])

finds_df=pd.DataFrame(finds, columns=['MRN','keywords_counts'])
finds_df
```

```python
output=pd.merge(merge,finds_df,on='MRN')
output=output[[col for col in output.columns if '_sch' not in col]]
output=output[[col for col in output.columns if 'explanation' not in col]]
output.columns=[col.replace('_me','') for col in output.columns]
output
```
