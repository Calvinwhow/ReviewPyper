```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

```python
def plot_accuracy_iterations(df, 
                             title='ReviewPyper Evaluation Accuracy',
                             filename=False):

    df=pd.melt(df, ['Timepoints'])
    palette = sns.color_palette('tab10')

    # Plotting
    plt.figure(figsize=(10, 6))
    sns.lineplot(data=df, x='Timepoints', y='value', hue='variable', style='variable', 
                markers=True,dashes=True, palette=palette, hue_order=None, style_order=None, 
                units=None, estimator='mean', errorbar=None, sort=True, err_style=None, err_kws=None,
                legend='full', ax=None)

    plt.legend(title='Model', loc='upper left')
    plt.xticks(rotation=45)
    plt.title(title)
    # plt.xlabel('Smooth')
    plt.ylabel('Accuracy')
    plt.tight_layout()
    sns.despine()
    plt.grid(False)
    if filename:
        plt.savefig(filename)
    plt.show()
```

```python

output_dir = '/Users/rm026/Documents/code/reviewpyper_testing/plots'

df=pd.DataFrame()
df['Timepoints']=['Initial Test',
            'Reduced\nNegatives',
            "Increased\nPositives",
            "Reduced\nCost",
            "Improved\nSpeed",
            "Parallelized\nQuestions",
            'Reduced\nFalse Positives',
            'Reduced\nFalse Negatives',
            'Reduced\nContext Window',]
df['GPT-4']=[.45,.61,.75,.75,.84,.84,.84,.91,.94,]
df['GPT-3.5']=[.4,.51,.51,.51,.75,.75,.75,.85,.92,]
```

```python
plot_accuracy_iterations(df, 
                         title='BARS Questions ReviewPyper Accuracy',
                         filename=os.path.join(output_dir, 'bars_reviewpyper_accuracy.png'))
```

```python
moca=pd.DataFrame({'Timepoints': ['Initial Test', 'Reduced\nFalse Positives'],
                   'GPT-4': [.85, 0.92],})
ccas=pd.DataFrame({'Timepoints': ['Initial Test', 'Reduced\nFalse Positives'],
                   'GPT-4': [.93, 0.99],})
cnrs=pd.DataFrame({'Timepoints': ['Initial Test', 'Reduced\nFalse Positives'],
                   'GPT-4': [.92, .98],})

plot_accuracy_iterations(moca, 
                         title='MOCA Evaluation Accuracy',
                         filename=os.path.join(output_dir, 'moca_reviewpyper_accuracy.png'))
plot_accuracy_iterations(ccas, 
                         title='CCAS Evaluation Accuracy',
                         filename=os.path.join(output_dir, 'ccas_reviewpyper_accuracy.png'))
plot_accuracy_iterations(cnrs, 
                         title='CNRS Evaluation Accuracy',
                         filename=os.path.join(output_dir, 'cnrs_reviewpyper_accuracy.png'))
```
